# Phase 5 - Targeting Policy and Business Impact

Renders `docs/policy_results.json` from `python -m uplift.policy`. Rupee figures use **illustrative assumptions** (cost per impression, value per visit), not Criteo data.

```
import json
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

results = json.loads(Path('../docs/policy_results.json').read_text())
print(results['assumptions'])
```

## Incremental outcomes vs budget (uplift vs response vs random ranking)

```
fig, ax = plt.subplots(figsize=(8, 5))
for name, s in results['strategies'].items():
    c = pd.DataFrame(s['curve'])
    ax.plot(c['fraction'], c['incremental_outcomes'], label=name)
ax.set_xlabel('fraction of users targeted')
ax.set_ylabel('estimated incremental visits')
ax.legend()
plt.show()
```

## Profit vs budget and the chosen operating point

```
fig, ax = plt.subplots(figsize=(8, 5))
for name, s in results['strategies'].items():
    c = pd.DataFrame(s['curve'])
    ax.plot(c['fraction'], c['profit'], label=name)
op = results['operating_point']
ax.axvline(op['fraction'], color='k', linestyle='--', label='uplift operating point')
ax.set_xlabel('fraction of users targeted')
ax.set_ylabel('profit (INR)')
ax.legend()
plt.show()
print(op)
```
